```python
# **Gradient Descent Algorithm using NumPy**

import numpy as np
```

```python
# Initalising  the intial values  
r = 8
s = 1
t = 4

print(f"The Intial values are r = {r}, s = {s}, t = {t}")

#Assigning learning rate and no of iterations
learning_rate = 0.01 
n_iterations = 100

for i in range(n_iterations):
    
    y =(r - 7)**3 + (4*s - 20)**2 + (2*s - 10)**2 + (4*(t**2/t) - 12)**2
    
    E = y**2 #Error function
    
    print(f"Iteration: {i} | r = {r} | s = {s} | t = {t} | Error = {E}")
    
    # Calculate the gradients (derivatives) of E with respect to r, s, and t
    
    dr = 3 * (r - 7)**2  # Derivative of E with respect to r
    ds = 10 * (4*s - 20)  # Derivative of E with respect to s
    dt = 8 * (4*t - 12) # Derivative of E with respect to t

    # Update the values with learing rate:
    r_new = r - learning_rate * dr
    s_new = s - learning_rate * ds
    t_new = t - learning_rate * dt
    #Assinging the new values
    r = r_new
    s = s_new
    t = t_new


print(f"The values are r = {r} s = {s} t = {t} and the final error is {E}")
```
